```python
%gui qt

import numpy as np
import pyqtgraph as pg
import pyqtgraph.opengl as gl
from pyqtgraph.Qt import QtCore
```

```python
def lorenz_step(state, sigma=10.0, rho=28.0, beta=8/3):
    x, y, z = state
    dx = sigma * (y - x)
    dy = x * (rho - z) - y
    dz = x * y - beta * z
    return np.array([dx, dy, dz])


def simulate_lorenz(n_steps=10000, dt=0.01):
    states = np.zeros((n_steps, 3))
    state = np.array([1.0, 1.0, 1.0])

    for i in range(n_steps):
        k1 = lorenz_step(state)
        k2 = lorenz_step(state + 0.5 * dt * k1)
        k3 = lorenz_step(state + 0.5 * dt * k2)
        k4 = lorenz_step(state + dt * k3)

        state = state + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        states[i] = state

    return states
```

```python
app = pg.mkQApp("Lorenz Attractor")

view = gl.GLViewWidget()
view.show()
view.setWindowTitle('Lorenz Attractor')

view.setCameraPosition(distance=80)
```

```python
grid = gl.GLGridItem()
grid.scale(5, 5, 5)
view.addItem(grid)
```

```python
states = simulate_lorenz(20000)
```

```python
curve = gl.GLLinePlotItem(
    pos=states,
    color=(1, 0, 0, 1),
    width=2,
    antialias=True
)

view.addItem(curve)
```

```python
ptr = 0
window = 2000  # number of points visible

def update():
    global ptr
    if ptr >= len(states):
        return

    start = max(0, ptr - window)
    curve.setData(pos=states[start:ptr])
    ptr += 10  # skip points → faster rendering

timer = QtCore.QTimer()
timer.timeout.connect(update)
timer.start(16)  # ~60 FPS
```

```python
import imageio

frames = []

for i in range(0, len(states), 20):
    curve.setData(pos=states[:i])
    img = view.grabFramebuffer()
    frames.append(img)

imageio.mimsave("lorenz.mp4", frames, fps=30)
```

```python
# -------------------------------
# 0. Jupyter Qt event loop
# -------------------------------
%gui qt

# -------------------------------
# 1. Imports
# -------------------------------
import numpy as np
import pyqtgraph as pg
import pyqtgraph.opengl as gl
from pyqtgraph.Qt import QtCore
import imageio

# -------------------------------
# 2. Lorenz System (RK4)
# -------------------------------
def lorenz_step(state, sigma=10.0, rho=28.0, beta=8/3):
    x, y, z = state
    dx = sigma * (y - x)
    dy = x * (rho - z) - y
    dz = x * y - beta * z
    return np.array([dx, dy, dz])


def simulate_lorenz(n_steps=20000, dt=0.01):
    states = np.zeros((n_steps, 3))
    state = np.array([1.0, 1.0, 1.0])

    for i in range(n_steps):
        k1 = lorenz_step(state)
        k2 = lorenz_step(state + 0.5 * dt * k1)
        k3 = lorenz_step(state + 0.5 * dt * k2)
        k4 = lorenz_step(state + dt * k3)

        state = state + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
        states[i] = state

    return states


# -------------------------------
# 3. Generate trajectory
# -------------------------------
states = simulate_lorenz()

# -------------------------------
# 4. Create 3D Viewer
# -------------------------------
app = pg.mkQApp("Lorenz Attractor")

view = gl.GLViewWidget()
view.show()
view.setWindowTitle('Lorenz Attractor')
view.setCameraPosition(distance=80)

# Grid
grid = gl.GLGridItem()
grid.scale(5, 5, 5)
view.addItem(grid)

# Trajectory line
curve = gl.GLLinePlotItem(
    pos=states[:1],
    color=(1, 0, 0, 1),
    width=2,
    antialias=True
)
view.addItem(curve)

# -------------------------------
# 5. Video Writer (streaming)
# -------------------------------
writer = imageio.get_writer("lorenz.mp4", fps=30, codec='libx264')

# -------------------------------
# 6. Animation + Capture Loop
# -------------------------------
ptr = 0
step = 20  # skip for speed

def update():
    global ptr

    if ptr >= len(states):
        writer.close()
        print("Video saved as lorenz.mp4")
        timer.stop()
        return

    # Update trajectory
    curve.setData(pos=states[:ptr])
    ptr += step

    # Capture frame
    img = view.grabFramebuffer().convertToFormat(4)  # RGBA
    ptr_bits = img.bits()
    ptr_bits.setsize(img.width() * img.height() * 4)

    frame = np.array(ptr_bits).reshape(img.height(), img.width(), 4)

    # Write frame
    writer.append_data(frame)


# -------------------------------
# 7. Timer (runs animation)
# -------------------------------
timer = QtCore.QTimer()
timer.timeout.connect(update)
timer.start(16)  # ~60 FPS
```
