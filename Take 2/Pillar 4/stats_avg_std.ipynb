```python
import json
import numpy as np

def compute_stats(json_file, key):
    with open(json_file, 'r') as f:
        data = json.load(f)

    values = np.array([entry[key] for entry in data if key in entry])

    stats = {
        "mean": round(float(np.mean(values)), 2),
        "std": round(float(np.std(values)), 2),
        "std_sample": round(float(np.std(values, ddof=1)), 2),
        "min": round(float(np.min(values)), 2),
        "max": round(float(np.max(values)), 2)
    }

    return stats



# Example usage
stats = compute_stats("stress_test_summary.json", "ITWAE")
stats
```
